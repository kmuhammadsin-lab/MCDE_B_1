# Task1:
Shopping Bill
Create a program that stores three items and their prices.

Requirements:
- Create a list containing three item names.
- Create another list containing their prices as integers or floats.
- Store the shop name as a string.
- Calculate and print the total price.
- Print all the information.

```python

# Shopping Bill

# Customer information
customer_name = "M.sinan"
shop_name = "Accessories"

# Store three shopping items
items = ["charger", "handsfree", "mike"]

# Store the price of each item
prices = [122, 444, 555]

# Calculate the total bill
total_bill = sum(prices)
print(total_bill)  # Complete this calculation

# Store bill information in a dictionary
bill = {
    "customer": customer_name,
    "shop": shop_name,
    "items": items,
    "prices": prices,
    "total": total_bill
}

# Display the shopping bill
print("----- SHOPPING BILL -----")
print("Customer:", customer_name)
print("Shop:", shop_name)
print("Items:", items)
print("Prices:", prices)
print("Total Bill: Rs.", total_bill)
```

- Store the item names in a list.
- Store your shop's name as a string.
- Calculate and print the total expense.
Example: You bought milk for Rs. 250, bread for Rs. 150, and eggs for Rs. 300.

```python

# Personal Expense Tracker

name = "Sinan"  # Enter your name
shop_name = "MSK"  # Enter the shop name

# Store the names of three items
items = ["charger", "mike", "handsfree"]

# Store the prices of the items
price1 = 122
price2 = 444
price3 = 5555

# Calculate the total expense
total_expense = price1 + price2 + price3
print("total expenses",total_expense) # Complete this calculation

# Display the details
print("Customer:", name)
print("Shop:", shop_name)
print("Items:", items)
print("Total Expense:", total_expense)
```

# Task 3
Food Delivery Order 🍔
Scenario: You are placing an order on a food delivery app.

Tasks:
- Store the restaurant name as a string.
- Store three food items in a list.
- Store the price of each item in another list.
- Store the delivery fee as a float.
- Calculate and print the total bill, including delivery.
Example: Burger Rs. 450, fries Rs. 200, and a drink Rs. 100, with a delivery fee of Rs. 80.

```python

# Food Delivery Order

restaurant_name = "ZRK"

# Store three food items
food_items = ["burger", "pizza", "sandwich"]

# Store the prices of the food items
prices = [12, 88, 99]

delivery_fee = 10.0

# Calculate the food cost and final bill
food_total = sum(prices)
  # Complete this calculation
final_bill = food_total + delivery_fee
print("final bill",final_bill)  # Include delivery fee

# Store order details in a dictionary
order = {
    "restaurant": restaurant_name,
    "items": food_items,
    "prices": prices,
    "delivery_fee": delivery_fee
}

# Display the order
print("Order Details:", order)
print("Food Total:", food_total)
print("Final Bill:", final_bill)
```
